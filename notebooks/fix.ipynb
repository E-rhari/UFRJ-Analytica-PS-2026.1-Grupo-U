```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import pickle
df = pd.read_csv("../datasets/BaseDPEvolucaoMensalCisp.csv", sep=";", decimal=",", encoding="latin1")

df.info()
```

```python
print(df["regiao"].unique())
```

```python
df = df[ df["regiao"].isin(["Capital"])]
df.info()
```

```python
df["risp"].value_counts()
```

```python
df[ df["risp"]==6]
```

```python
## esses dados estão com a risp sendo 6, porem a regiao capital e cisp 16 se referem a barra da tijuca, pertencente a risp 2. 
## como sao poucos dados, assumimos que se trata de um erro de preenchimento, e trocaamos para o valor correto

df.loc[:,"risp"] = df.loc[:,"risp"].replace(6,2)
df["risp"].value_counts()
```

```python
df[ df["risp"]==5]
```

```python
# mesma logica da anterior, porém aqui a risp correta é 3
df.loc[:,"risp"] = df.loc[:,"risp"].replace(5,3)
df["risp"].value_counts()
```

```python
df["mes_ano"] = pd.to_datetime(
    df.loc[:,'mes_ano'].str.replace('m', '-', regex=False),
    format='%Y-%m'
)
```

```python
df.info()
```

```python
df = df[df["ano"] >=2014]
```

```python
df.to_pickle('../datasets/dataframe.pkl')
```
